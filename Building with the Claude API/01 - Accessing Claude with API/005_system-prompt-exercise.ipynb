```python
# Create an API client
from anthropic import Anthropic

client = Anthropic()
model = "claude-sonnet-5"
```

```python
def add_user_message(messages, text):
    user_message = {"role": "user", "content": text}
    messages.append(user_message)


def add_assistant_message(messages, text):
    assistant_message = {"role": "assistant", "content": text}
    messages.append(assistant_message)

def chat(messages, system=None):

    params = {
        "model": model,
        "max_tokens": 1000,
        "messages": messages,
    }

    if system is not None:
        params["system"] = system

    response = client.messages.create(**params)
    return "".join(block.text for block in response.content if block.type == "text")
```

```python
messages=[]
system_prompt = "You are a Python engineer that writes extremely concise Python code."
add_user_message(messages, "Write a Python function that checks a string for duplicate characters.")
answer = chat(messages,system_prompt)
print(answer)
```
